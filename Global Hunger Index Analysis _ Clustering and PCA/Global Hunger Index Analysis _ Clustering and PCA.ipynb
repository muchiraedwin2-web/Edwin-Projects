```python
# Create new data frame for clustering
df_GHI2022['ufm_rate']=df_ufm_rate['2022']
# df_GHI2022['prop_un']=df_prop_un['2022']
df_GHI2022['preva_st_u5']=df_preva_st_u5['2022']
df_GHI2022['preva_wa_u5']=df_preva_wa_u5['2022']
```

```python
df_clus=df_GHI2022.drop(['2000','2007','2014'], axis=1)
df_clus
```

```python
# Drop NaN
df_clus= df_clus.dropna()
```

```python
df_clus2= df_clus.drop('Country', axis=1)
```

```python
# Standardization
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
df_sc = sc.fit_transform(df_clus2)
df_sc = pd.DataFrame(df_sc, columns=df_clus2.columns)
```

```python
# Modeling
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# Elbow method
distortions = []

for i in range(1, 11):
    km = KMeans(
        n_clusters=i,
        init='k-means++',
        n_init=10,
        max_iter=300,
        random_state=0
    )

    km.fit(df_sc)
    distortions.append(km.inertia_)

plt.plot(range(1, 11), distortions, marker='o')
plt.xlabel('Number of clusters')
plt.ylabel('Distortion')
plt.title('Elbow Method')
plt.show()
```

```python
# Cluster labeling
model = KMeans(n_clusters=4, random_state=1)
model.fit(df_sc)
```

```python
cluster = model.labels_
```

```python
df_clus['Cluster']=cluster
```

```python
df_clus.groupby('Cluster').mean(numeric_only=True).style.bar(axis=0)
```

```python
# Boxplot by clusters
col=['2022', 'Absolute change since 2014', '% change since 2014', 'ufm_rate', 'preva_st_u5', 'preva_wa_u5']
```

```python
fig = plt.figure(figsize=(10,10))

for i in range(len(col)):
    plt.subplot(3,3, i+1)
    plt.title(col[i])
    sns.boxplot(data=df_clus, y=df_clus[col[i]], x=df_clus['Cluster'], palette='Set1')
    
plt.tight_layout()
plt.show()
```

```python
# PCA
# Modeling
from sklearn.decomposition import PCA
pca = PCA(n_components=2, random_state=1)

pca.fit(df_sc)
feature = pca.transform(df_sc)
```

```python
df_clus['PCA1']=feature[:, 0]
df_clus['PCA2']=feature[:, 1]
```

```python
df_clus.groupby('Cluster')[['PCA1', 'PCA2']].mean().T.style.bar(axis=1)
```

```python
# Scatterplot by components and clusters
pd.DataFrame(pca.components_, columns=df_sc.columns, index=['PCA1','PCA2'])
```

```python
fig = plt.figure(figsize=(12,6))
ax = fig.add_subplot()

scatter = ax.scatter(df_clus['PCA1'], df_clus['PCA2'], alpha=0.8, c=cluster)
ax.set_xlabel('principal component 1')
ax.set_ylabel('principal component 2')
plt.legend(handles=scatter.legend_elements()[0], labels=['Cluster0','Cluster1','Cluster2','Cluster3'],
           title="Cluster",loc='upper left', bbox_to_anchor=(1, 1))
plt.show()
```
